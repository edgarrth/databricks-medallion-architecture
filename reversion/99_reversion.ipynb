# 99 - Reversión

Elimina tablas y schemas del proyecto. Usar solo para limpieza controlada.


```python
dbutils.widgets.text("catalog_name", "fintech_lakehouse")
catalog_name = dbutils.widgets.get("catalog_name")
```

```python
for schema_name in ["gold", "silver", "bronze"]:
    tables = spark.sql(f"SHOW TABLES IN {catalog_name}.{schema_name}").collect()
    for row in tables:
        spark.sql(f"DROP TABLE IF EXISTS {catalog_name}.{schema_name}.{row.tableName}")
    spark.sql(f"DROP SCHEMA IF EXISTS {catalog_name}.{schema_name}")

# Descomentar solo si deseas eliminar el catálogo completo.
# spark.sql(f"DROP CATALOG IF EXISTS {catalog_name}")
```
